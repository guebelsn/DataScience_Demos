## 1. Import the Data 8 tools

Berkeley Data 8 commonly uses `from datascience import *`. This gives us the `Table` tools we'll use to read and explore data.

We will also import NumPy (`np`) for one simple calculation.

```python
#please don't change this cell!
from datascience import *
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')
```

## 2. Read in our first table: plant growth

Our CSV file contains measurements from 12 fictional plants.

- Each **row** represents one plant.
- Each **column** represents something measured about that plant.

`Table.read_table()` reads the CSV file and creates a Data 8 table.

```python
plants = Table.read_table("plant_growth.csv")
```

### Take a first look

Typing the table's name displays it. `show()` lets us choose how many rows to display.

```python
plants.show(5)
```

Let's inspect the table's column labels and its number of rows.

```python
print("Column labels:", plants.labels)
print("Number of rows:", plants.num_rows)
```

### Choose columns

`select()` creates a table containing only the columns we name. This is useful when a dataset has many measurements but we only want to focus on a few.

```python
plants.select("Plant_ID", "Height_cm", "Leaf_Count")
```

## 3. Make a scatter plot: light and plant height

**Biology question:** Do plants that receive more hours of light tend to grow taller?

In a scatter plot, each dot represents one plant. The first column named in `scatter()` goes on the x-axis, and the second goes on the y-axis.

```python
plants.scatter("Light_Hours", "Height_cm")
```

**Discuss:** What pattern do you notice? Do plants with more light tend to be taller in this example?

A pattern in this small fictional dataset does not prove that light alone caused the differences in height.

## 4. A second scatter plot: light and leaf count

We can use the same table to ask another question.

```python
plants.scatter("Light_Hours", "Leaf_Count")
```

**Discuss:** Does the number of leaves appear to change as light exposure increases?

## 5. Read a second table: bacterial growth

This fictional dataset records optical density at 600 nm (`OD600`) over time.

A higher OD600 reading often indicates a more turbid culture and can be used as an indirect measure of growth. It is not a direct cell count.

```python
bacteria = Table.read_table("bacterial_growth.csv")
bacteria.show()
```

### Make a line plot

Because the measurements are ordered over time, a line plot is a useful way to show the trend.

```python
bacteria.plot("Time_hours", "OD600")
```

## 6. Calculate an average

Data 8 tables let us access a column by name using `column()`.

Here, we'll use NumPy's `mean()` function to calculate the average plant height.

```python
heights = plants.column("Height_cm")
average_height = np.mean(heights)

print("Average plant height:", round(average_height, 2), "cm")
```

## 8. Quick recap

| What we wanted to do | Data 8 code |
|---|---|
| Read a CSV file | `Table.read_table("file.csv")` |
| Display a table | `table.show()` |
| See column names | `table.labels` |
| Count rows | `table.num_rows` |
| Keep selected columns | `table.select("A", "B")` |
| Get one column | `table.column("A")` |
| Make a scatter plot | `table.scatter("x", "y")` |
| Make a line plot | `table.plot("x", "y")` |
| Calculate an average | `np.mean(...)` |

**Big idea:** Data 8's `Table` makes it straightforward to read, inspect, and visualize data. We can use those tools to explore biological questions.
