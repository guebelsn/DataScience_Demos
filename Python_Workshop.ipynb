Now, lets try loading in a data set!
- 
- Today we have some cal fresh data to look at! The file we will load is called CalFreshAnnual.csv
    - This data came from:  https://www.cdss.ca.gov/inforesources/data-portal/research-and-data/calfresh-data-dashboard
- in order to load our data table, we can use this function: Table().read_table('CalFreshAnnual.cvs')
- In order to save the table, we need to assign it to a variable! Here, we'll call this variable 'fresh'

```python
# Lets use the Table() function to load in our table and then look at it!
fresh = Table().read_table('CalFreshAnnual.csv')
fresh
```

WOW! That's a lot of numbers!!
- 
Now lets use python to visualize some of this data using a scatter plot!!
- First, lets narrow down the table and just look at the column for the year, the column representing the amount of money that was distributed statewide through the calfresh program and the column represented the amount districbuted in just Santa Barbara. 
- In order to make our knew reduced table, we can use the name of our table (which in this case is fresh) together with .select() and pass in the names of the columns we are interested in. Lets save this new table to a variable called new_table.


```python
# Let's focus in and look at a couple of columns and save the new table in a new variable name called new_table
new_table = fresh.select('Year of Calendar Year', 'Statewide', 'Santa Barbara')
new_table
```

```python
#JUST RUN THIS CELL!! Please don't change it. It is just to clean up the data a little bit!
formatter = NumberFormatter()
for i in range(1, new_table.num_columns):
    new_table = new_table.with_column(new_table.labels[i], new_table.apply(formatter.convert_value,i))
```

Maybe Graphing this data will make it easier to understand!
-
- to make a line plot, we can use the .plot() method and pass in the column names that we want represented on the x-axis and y-axis.
- Lets make 2 graphs, one looking at how the statewide calfresh funding changed over time and one looking at how the funding distributed just in Santa Barbara changed!

```python
new_table.plot('Year of Calendar Year', 'Statewide')
new_table.plot('Year of Calendar Year', 'Santa Barbara')
```
